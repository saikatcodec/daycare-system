## Check Dataset
Check the dataset is balanced or not

```python
import os
from pathlib import Path
```

##### How many no of adult and children classes present?

```python
classes = {}

annot_path = './datasets/child_vs_adult/train/labels'
annot_files = Path(annot_path).glob('*.txt')

for annot_file in annot_files:
    with open(annot_file, 'r') as file:
        for line in file:
            names = line.split(' ')
            
            if names[0] == '0':
                classes['adult'] = classes.get('adult', 0) + 1
            elif names[0] == '1':
                classes['child'] = classes.get('child', 0) + 1
```

```python
classes
```

### Conclusion
The dataset is not balanced

### Check labels and images file

```python
## Read all the images
file_path = './datasets/child_vs_adult/valid'
img_ext = ['*.jpg', '*.jpeg', '*.png']

images_path = []
for ext in img_ext:
    images_path.extend(Path(file_path + '/images').glob(ext))

len(images_path)
```

```python
Path(images_path[10]).stem, Path(images_path[10])
```

```python
no_valid = 0
label_path = os.path.join(file_path, 'labels')
for img_path in images_path:
    annot_path = Path(f'{os.path.join(label_path, Path(img_path).stem)}.txt')
    
    if annot_path.exists():
        no_valid += 1
    else:
        print(f'Path is not found for {annot_path}')

print(f'All images: {len(images_path)}\t valid annotation: {no_valid}')
```

### Conclusion
The dataset is clean
