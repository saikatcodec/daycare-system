# Train the model

```python
from ultralytics import YOLO

# pretrained model
model = YOLO('yolo12l.pt')

# train the model
results = model.train(
    data='datasets/behavioral_v3.0/data.yaml',
    epochs=500,
    imgsz=640,
    batch=-1,
    workers=10,
    patience=30,
    plots=True,
    cache='disk',
    project='Intellgi',
    name='action_classify'
)
```
