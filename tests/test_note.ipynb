```python
import cv2
import numpy as np
from cv2.typing import MatLike

def draw_bounding_box(image: MatLike, name: str, bbox: list, color: tuple[int, int, int]) -> MatLike:
    # Draw a bounding box for the prediction
    cv2.rectangle(
        image,
        (int(bbox[0]), int(bbox[1])),
        (int(bbox[2]), int(bbox[3])),
        color=color,
        thickness=2
    )
    
    # Constant
    FONT_FACE = cv2.FONT_HERSHEY_PLAIN
    FONT_SCALE = 1.0
    THICKNESS = 2
    # Get text size
    (text_width, text_height), baseline = cv2.getTextSize(
        text=name,
        fontFace=FONT_FACE,
        fontScale=FONT_SCALE,
        thickness=THICKNESS
    )
    # Calculate the box position based on text size
    text_box_min = int(bbox[0]), int(bbox[1])
    text_box_max = int(bbox[0] + text_width + baseline), int(bbox[1] + text_height + baseline)
    # Draw a rectangle according to the text size
    cv2.rectangle(
        image,
        text_box_min,
        text_box_max,
        color=color,
        thickness=-1
    )
    # Put text into the box
    text_x = int(text_box_min[0] + baseline // 2)
    text_y = int(text_box_max[1] - baseline // 2)
    cv2.putText(
        image,
        name,
        (text_x, text_y),
        fontFace=FONT_FACE,
        fontScale=FONT_SCALE,
        thickness=THICKNESS,
        color=(255, 255, 255)
    )

    return image

# Image setup (create a black image)
img = np.ones((400, 800, 3), dtype=np.uint8)

# Text parameters
text = "Hello, OpenCV!"
font = cv2.FONT_HERSHEY_PLAIN
fontScale = 1
thickness = 2

# Get the text size and baseline
(text_width, text_height), baseline = cv2.getTextSize(text, font, fontScale, thickness)

print(f"Text Width: {text_width}px")
print(f"Text Height: {text_height}px")
print(f"Baseline: {baseline}px")

img = draw_bounding_box(
    img,
    text,
    [50, 40, 400, 200],
    (234, 23, 23)
)

# Display the image (optional)
cv2.imshow("Image with Text", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
